```python
import numpy as np
from scipy.integrate import odeint
```

```python
def simple_ode(x, t):
    return [
        -0.05*x[0],
        -x[1] + x[0]**2
    ]
```

```python
dt = 0.02
t = np.arange(0, 100, dt)
```

```python
x =[]
x_next = []
```

```python
for x0_0 in np.linspace(-1, 1, 10):
    for x0_1 in np.linspace(-1, 1, 10):
        x0 = np.array([x0_0, x0_1])
        integral = odeint(simple_ode, x0, t)
        x.append(integral[:-1 , :])
        x_next.append(integral[1:, :]) 
```

```python
x = np.vstack(x)
x_next = np.vstack(x_next)
```
